```python
arr = rng.random(10_000)

start = time.time()
loop_result = [ x * 2 for x in arr]
loop_time = time.time() - start

start =  time.time()
vec_result = arr * 2
vec_time = time.time() - start

print(f"loop: {loop_time:.5f}s, vectorized: {vec_time:.5f}s, speedup: {loop_time/vec_time:.1f}x")

%timeit loop_result
%timeit vec_result
```
