```python
import pandas as pd

df = pd.DataFrame({
    "name": ["A", "B", "C", "D"],
    "age": [20, 25, 30, 35],
    "score": [80, 65, 90, 70]
}, index=["a", "b", "c", "d"])
df
```

```python
df.iloc[0:2, 0:2]
```

```python
df[df["age"] < 30]
```

```python
df.query("age >= 30 and score >70")
```

```python
df.sort_values("score")
```

```python
#merge
df2 = pd.DataFrame({
    "name": ["E", "F", "G", "H"],
    "age": [1500, 75, 60, 40],
    "score": [80, 65, 90, 70]
}, index=["a", "b", "c", "d"])
df_mrg = pd.merge(df,df2,how="outer")
df_mrg
```

```python
#concat
cat = pd.concat([df,df2],axis=0)
cat
```
